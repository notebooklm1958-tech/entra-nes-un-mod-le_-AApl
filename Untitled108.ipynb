### Étape 1 : Téléchargement et préparation des données historiques

**Description :** Cette cellule charge les bibliothèques indispensables (`pandas`, `yfinance`, `scikit-learn`, `matplotlib`), vous redirige vers TradingView pour que vous puissiez trouver le symbole exact de l'actif souhaité, puis télécharge ses données boursières via l'API Yahoo Finance.

**Interprétation des résultats :** La cellule affiche le nombre total de lignes récupérées et un aperçu des dernières lignes pour valider le bon formatage de l'index temporel et des colonnes de prix (Clôture, Plus Haut, Plus Bas) et de Volume.

```python
# ============================================================
# PROJET : PREMIER MODELE DE MACHINE LEARNING SUR AAPL
# Objectif : estimer si AAPL montera le jour suivant
# ============================================================

import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# ETAPE 1 : TELECHARGER LES DONNEES
# ------------------------------------------------------------

print("--- RECHERCHE DU SYMBOLE SUR TRADINGVIEW ---")
print("Si vous n'êtes pas sûr du symbole (ticker), vous pouvez faire votre choix sur TradingView :")
print("👉 https://www.tradingview.com/symbols/\n")

# Demande interactive à l'utilisateur
ticker_input = input("Saisissez le ticker trouvé sur TradingView (ex: AAPL, TSLA, BTC-USD) [Défaut: AAPL] : ").strip()
ticker = ticker_input if ticker_input != "" else "AAPL"

period_input = input("Saisir la période (ex: 1mo, 3mo, 6mo, 1y, 2y, 5y, max) [Défaut: 2y] : ").strip()
period = period_input if period_input != "" else "2y"

interval_input = input("Saisir l'intervalle (ex: 1m, 5m, 15m, 1h, 1d) [Défaut: 1h] : ").strip()
interval = interval_input if interval_input != "" else "1h"

print(f"\nTéléchargement de {ticker} avec une période de {period} et un intervalle de {interval}...")

data = yf.download(
    ticker,
    period=period,
    interval=interval,
    auto_adjust=True,
    progress=False
)

# Compatibilité avec les colonnes MultiIndex de yfinance
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

if data.empty:
    raise ValueError(
        f"Aucune donnée reçue pour {ticker}. Vérifie ta connexion Internet ou les paramètres saisis."
    )

data = data[["Close", "High", "Low", "Volume"]].copy()
data = data.dropna()

print(f"Nombre de lignes téléchargées pour {ticker} :", len(data))
display(data.tail())
```

### Étape 2 : Création des variables d'apprentissage (Features)

**Description :** Dans cette étape, nous construisons les indicateurs techniques mathématiques qui serviront d'indices à notre modèle de Machine Learning pour prédire l'avenir : le rendement, des moyennes mobiles à 50 et 100 périodes, le ratio de distance aux moyennes mobiles, la volatilité historique, le RSI (Indice de Force Relative) et la variation des volumes.

**Interprétation des résultats :** L'aperçu final affiche les dernières lignes des variables créées. Ces indicateurs agrègent le bruit quotidien pour mettre en valeur les tendances de prix et de dynamique (momentum).

```python
# ------------------------------------------------------------
# ETAPE 2 : CREER LES VARIABLES D'APPRENTISSAGE
# ------------------------------------------------------------

# Rendement journalier
data["Return"] = data["Close"].pct_change()

# Moyennes mobiles (Mise à jour pour MA_50 et MA_100)
data["MA_50"] = data["Close"].rolling(50).mean()
data["MA_100"] = data["Close"].rolling(100).mean()

# Position du cours par rapport à la moyenne mobile à long terme
data["MA_Ratio"] = data["Close"] / data["MA_100"] - 1

# Volatilité sur 50 séances
data["Volatility"] = data["Return"].rolling(50).std()

# RSI sur 14 séances, calculé avec des moyennes simples
delta = data["Close"].diff()
gains = delta.clip(lower=0)
losses = -delta.clip(upper=0)

avg_gain = gains.rolling(14).mean()
avg_loss = losses.rolling(14).mean()

rs = avg_gain / avg_loss.replace(0, np.nan)
data["RSI"] = 100 - (100 / (1 + rs))

# Variation du volume
data["Volume_Change"] = data["Volume"].pct_change()

display(data[["Close", "Return", "MA_50", "MA_100", "MA_Ratio", "Volatility", "RSI", "Volume_Change"]].tail())
```

### Étape 3 : Définition de l'objectif à prédire (Target) et sélection finale

**Description :** Nous créons ici la cible (`Target`) que l'algorithme doit apprendre à prédire. Il s'agit d'une variable binaire : `1` si le cours monte le lendemain, ou `0` s'il baisse ou stagne. Nous nettoyons ensuite l'ensemble des données pour supprimer les valeurs manquantes (générées par les calculs de moyennes mobiles) et isolons nos variables d'entrée `X` de la cible `y`.

**Interprétation des résultats :** Le programme affiche le nombre final d'exemples exploitables pour l'apprentissage ainsi que la liste des indicateurs qui serviront à l'entraînement.

```python
# ------------------------------------------------------------
# ETAPE 3 : DEFINIR CE QUE LE MODELE DOIT APPRENDRE
# ------------------------------------------------------------

# 1 = le cours de clôture suivant est supérieur au cours actuel
# 0 = le cours suivant est inférieur ou égal au cours actuel

next_return = data["Close"].shift(-1) / data["Close"] - 1

data["Target"] = np.where(
    next_return.isna(),
    np.nan,
    (next_return > 0).astype(int)
)

# Variables que le modèle utilisera pour apprendre
features = [
    "Return",
    "MA_Ratio",
    "Volatility",
    "RSI",
    "Volume_Change"
]

# Conserver les lignes complètes
dataset = data[features + ["Target"]].replace(
    [np.inf, -np.inf], np.nan
).dropna()

X = dataset[features]
y = dataset["Target"].astype(int)

print("\nNombre d'exemples d'apprentissage :", len(dataset))
print("Variables utilisées :", features)
```

### Étape 4 : Division des données (Train / Test Split)

**Description :** Pour évaluer de manière réaliste notre modèle, nous séparons les données en deux ensembles chronologiques distincts : 80 % pour l'entraînement (le passé) et 20 % pour le test (le futur le plus récent). Il est crucial de ne pas mélanger les dates pour éviter tout biais d'anticipation temporelle (look-ahead bias).

**Interprétation des résultats :** Cette séparation nous montre le nombre précis d'exemples d'apprentissage historique d'un côté, et le nombre d'exemples sur lesquels nous allons simuler les prédictions futures de l'autre.

```python
# ------------------------------------------------------------
# ETAPE 4 : SEPARER LES DONNEES DANS L'ORDRE CHRONOLOGIQUE
# ------------------------------------------------------------

# Ne pas mélanger les dates : le passé sert à apprendre,
# la période récente sert à tester.

split = int(len(dataset) * 0.80)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("\nExemples d'entraînement :", len(X_train))
print("Exemples de test :", len(X_test))
```

### Étape 5 : Création du Pipeline et entraînement du modèle

**Description :** Nous assemblons un pipeline qui applique d'abord une normalisation standard aux données (pour que toutes les variables aient la même échelle) puis entraîne un modèle de Régression Logistique. Ce modèle va analyser les relations statistiques entre nos indicateurs techniques et la hausse ou la baisse du jour suivant.

**Interprétation des résultats :** Le message de succès confirme que l'algorithme a convergé et a appris à ajuster ses coefficients mathématiques en fonction de l'historique d'entraînement.

```python
# ------------------------------------------------------------
# ETAPE 5 : CREER ET ENTRAINER LE MODELE
# ------------------------------------------------------------

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000)
)

# L'algorithme apprend les relations entre les variables
# et la direction observée du cours suivant.
model.fit(X_train, y_train)

print("\nModèle entraîné avec succès !")
```

### Étape 6 : Évaluation des performances du modèle

**Description :** Cette cellule teste le modèle sur la période récente (les 20 % de données exclues lors de l'apprentissage) pour mesurer son efficacité réelle. Elle génère le score global de précision (accuracy), un rapport détaillé et la matrice de confusion.

**Interprétation des résultats :**
* L'**accuracy** indique le pourcentage global de bonnes directions estimées (généralement situé légèrement au-dessus de 50 % sur les marchés financiers).
* Le **rapport de classification** montre la précision et le rappel par classe.
* La **matrice de confusion** détaille le nombre exact de vrais positifs, vrais négatifs, faux positifs et faux négatifs commis par l'algorithme.

```python
# ------------------------------------------------------------
# ETAPE 6 : EVALUER LE MODELE SUR LES DONNEES DE TEST
# ------------------------------------------------------------

predictions = model.predict(X_test)

accuracy = accuracy_score(y_test, predictions)

print("\nPrécision globale (accuracy) :",
      round(accuracy * 100, 2), "%")

print("\nRapport de classification :")
print(classification_report(
    y_test,
    predictions,
    labels=[0, 1],
    target_names=["Baisse ou stabilité", "Hausse"],
    zero_division=0
))

print("Matrice de confusion :")
print(confusion_matrix(y_test, predictions))
```

### Étape 7 : Estimation de la probabilité pour la prochaine séance

**Description :** Nous isolons les indicateurs de la toute dernière journée disponible dans notre base de données afin de faire une projection en temps réel. Le modèle calcule la probabilité de hausse par rapport à la probabilité de baisse ou stagnation pour la prochaine séance boursière.

**Interprétation des résultats :** Cette étape donne un résultat opérationnel direct. Elle affiche la date de la dernière clôture connue, son prix, ainsi que la probabilité estimée en pourcentage pour le mouvement de l'action lors de la séance suivante.

```python
# ------------------------------------------------------------
# ETAPE 7 : ESTIMER LA PROBABILITE POUR LA PROCHAINE SEANCE
# ------------------------------------------------------------

# Utiliser les indicateurs de la dernière séance disponible.
# Cette ligne n'a pas besoin de connaître le rendement futur.

latest_features = data[features].replace(
    [np.inf, -np.inf], np.nan
).dropna().iloc[[-1]]

if latest_features.empty:
    raise ValueError("Impossible de calculer les indicateurs récents.")

probabilities = model.predict_proba(latest_features)[0]

prob_baisse = probabilities[0]
prob_hausse = probabilities[1]

print("\nDernière date disponible :", data.index[-1].date())
print("Dernier cours de clôture :", round(data["Close"].iloc[-1], 2))
print("Probabilité estimée de baisse/stabilité :",
      round(prob_baisse * 100, 2), "%")
print("Probabilité estimée de hausse :",
      round(prob_hausse * 100, 2), "%")
```

### Étape 8 : Visualisation du cours et des moyennes mobiles

**Description :** Pour conclure, nous traçons un graphique montrant l'évolution historique du cours de l'actif superposée avec ses moyennes mobiles à 50 et 100 périodes afin de pouvoir analyser visuellement les tendances et les signaux de croisements.

**Interprétation des résultats :** Le graphique permet de repérer facilement si le marché est dans une phase haussière (lorsque le cours et la moyenne à 50 sont au-dessus de la moyenne à 100) ou dans une phase de consolidation/baisse.

```python
# ------------------------------------------------------------
# ETAPE 8 : VISUALISER LE COURS ET LES MOYENNES MOBILES
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(data.index, data["Close"], label=f"Cours {ticker}")
plt.plot(data.index, data["MA_50"], label="MA 50")
plt.plot(data.index, data["MA_100"], label="MA 100")

plt.title(f"{ticker} - Cours et moyennes mobiles (50 & 100)")
plt.xlabel("Date")
plt.ylabel("Prix (USD)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
